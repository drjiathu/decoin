```python
import requests
from bs4 import BeautifulSoup

def get_binance_subfolders():
    base_url = "https://data.binance.vision"
    path = "data/futures/um/daily/trades/"
    url = f"{base_url}/?prefix={path}"
    
    try:
        # 发送HTTP请求
        response = requests.get(url)
        response.raise_for_status()
        
        # 解析HTML内容
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # 查找所有链接（包含子文件夹）
        subfolders = []
        for link in soup.find_all('a'):
            href = link.get('href')
            if href and href.startswith(f"?prefix={path}") and href.endswith('/'):
                # 提取子文件夹名称（交易对名称）
                folder_name = href.split(path)[1].rstrip('/')
                if folder_name and folder_name != path.split('/')[-2]:
                    subfolders.append(folder_name)
        
        return sorted(subfolders)
    
    except Exception as e:
        print(f"Error: {e}")
        return []

# 获取并打印结果
trading_pairs = get_binance_subfolders()
print(f"Found {len(trading_pairs)} trading pairs:")
for pair in trading_pairs:
    print(pair)
```
