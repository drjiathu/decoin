```python
import datetime
from zoneinfo import ZoneInfo

from app.api.okx import fetch_historical_candles

end_ts = int(datetime.datetime(2025, 2, 8, tzinfo=ZoneInfo("UTC")).timestamp() * 1000)
start_ts = int(datetime.datetime(2025, 1, 1, tzinfo=ZoneInfo("UTC")).timestamp() * 1000)

df = fetch_historical_candles("BTC-USDT-SWAP", bar="4H", start_ts=start_ts-1, end_ts=end_ts+1)
assert (df["ts"].iat[0].value == start_ts * 1e6) & (df["ts"].iat[-1].value == end_ts * 1e6)
```

```python
import datetime
from zoneinfo import ZoneInfo

from app.api.okx import fetch_historical_funding_rate


start_ts = int(datetime.datetime(2024, 11, 1, tzinfo=ZoneInfo("UTC")).timestamp() * 1000)
end_ts = int(datetime.datetime(2024, 12 ,31, tzinfo=ZoneInfo("UTC")).timestamp() * 1000)
funding_rate_df = fetch_historical_funding_rate("BTC-USDT-SWAP", start_ts-1, end_ts+1)
s = funding_rate_df["fundingTime"]
assert (s.iat[0].value == start_ts * 1_000_000) & (s.iat[-1].value == end_ts * 1_000_000)
```
