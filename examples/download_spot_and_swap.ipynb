# Download spot trades, swap trades and funding rates from OKX

OKX provide historical data for spot trades, swap trades and funding rates.
The data is available for download in ZIP compressed CSV format from the OKX website https://www.okx.com/data-download.
The data is downloaded from monthly directories. The deepest ditrctory is the date of the data which is based on Hongkong time.

Downloading functions are categorized into two:

1. Downloading all data from 2021-10-01 to today
2. Downloading daily update data



## 1. Download all historical spot trades, swap trades and funding rates

## 1.1 Download all historical spot trades

```python
import datetime
from app.api.okx import download_daily_trades

download_daily_trades(date=datetime.date(2025, 6, 4))
```

```python
import datetime
from zoneinfo import ZoneInfo

import pandas as pd
from tqdm.notebook import tqdm

from app.api.okx import download_daily_funding_rates

end_date = datetime.datetime.now(tz=ZoneInfo("UTC")).date() - datetime.timedelta(days=1)
start_date = datetime.date(2021, 10, 1)

pbar = tqdm(pd.date_range(start=start_date, end=end_date))
for date in pbar:
    pbar.set_description(desc=f"Downloading {date.strftime("%Y%m%d")} swaprate")
    download_daily_funding_rates(date)
```
